```python
import cv2 
import numpy as np
import matplotlib.pylab as plt
%matplotlib inline
```

```python
def load_img():
    img = cv2.imread('Computer-Vision-with-Python/DATA/bricks.jpg').astype(np.float32)/255
    img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    return img
```

```python
#load_img()
```

```python
def display_img(img):
    plt.figure(figsize=(17,17))
    plt.imshow(img)
```

```python
i = load_img()
display_img(i)
```

```python
gamma = 2 #if gamma > 1 the results will be darker , if gamma < 1 the result will be lighter
```

```python
result = np.power(i,gamma)
```

```python
display_img(result)
```

```python
img = load_img()
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(img,'bricks',(10,600),font,10,(255,0,0),4)
```

```python
display_img(img)
```

```python
kernal = np.ones(shape=(5,5),dtype=np.float32)/25
```

```python
kernal
```

```python
dst = cv2.filter2D(img,-1,kernal)
display_img(dst)
```
